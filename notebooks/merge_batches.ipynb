```python
# code to merge batches

from pathlib import Path

BATCHES_FOLDER = Path('.../path/to/batches') # Replace with the actual path to your batches folder

files = list(BATCHES_FOLDER.glob('*.jsonl.gz'))
print(f'Found {len(files)} files in folder')
```

```python
import gzip
import json
import os
from tqdm import tqdm
import psutil

X = len(files) // 10  # a cada quantos arquivos reportar
total_items = 0
process = psutil.Process(os.getpid())

os.makedirs('results', exist_ok=True)

with open('results/whosampled-brainz.jsonl', 'w', encoding='utf-8') as out_f:
    for i, file in enumerate(tqdm(files), start=1):
        with gzip.open(file, 'rt', encoding='utf-8') as f:
            for line in f:
                item = json.loads(line)
                out_f.write(json.dumps(item, ensure_ascii=False) + '\n')
                total_items += 1

        if i % X == 0:
            rss_mb = process.memory_info().rss / 1e6
            tqdm.write(f"[{i}/{len(files)}] {total_items} itens escritos, RSS: {rss_mb:.2f} MB")
```
